Импорт основных библиотек


```python
import json
import os

import requests
from dotenv import load_dotenv

load_dotenv()
```

Ввод названия фильма; ключ OMDb берётся из `.env` (переменная `OMDB_API_KEY`).

```python
title = input("Введите фильм для поиска: ").strip()
api_key = os.getenv("OMDB_API_KEY")
if not api_key:
    raise RuntimeError("В .env нет OMDB_API_KEY (или файл не найден — запускайте из корня проекта).")
```

Отправка запроса

```python
try:
    response = requests.get(
        "http://www.omdbapi.com/",
        params={"t": title, "apikey": api_key},
        timeout=10,
    )
    response.raise_for_status()
except requests.RequestException as e:
    raise RuntimeError(f"Ошибка запроса к OMDb: {e}") from e

try:
    json_data = response.json()
except ValueError as e:
    raise RuntimeError("OMDb вернул ответ, который не удалось разобрать как JSON") from e

if json_data.get("Response") == "False":
    raise RuntimeError(json_data.get("Error", "Неизвестная ошибка OMDb"))
```

Вывод полного JSON (уже разобран в ячейке с запросом)

```python
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

Достаем нужные поля

```python
movie_title = json_data.get("Title")
year = json_data.get("Year")
genre = json_data.get("Genre")
director = json_data.get("Director")
imdb_rating = json_data.get("imdbRating")

print(f"Название: {movie_title}")
print(f"Год: {year}")
print(f"Жанр: {genre}")
print(f"Режиссёр: {director}")
print(f"Рейтинг IMDb: {imdb_rating}")
```

Актёры и краткая сводка

```python
actors = json_data.get("Actors")
print(f"Актёры: {actors}")

print(f"Фильм '{movie_title}' ({year}) — жанр: {genre}. Режиссёр: {director}.")
```
